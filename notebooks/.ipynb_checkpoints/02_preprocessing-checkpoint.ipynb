# Document preprocessing for OCR

This notebook builds a deterministic preprocessing workflow for a randomly selected image under `data/images`. It keeps the original image untouched and produces two OCR candidates: normalized grayscale and adaptive binary.

The workflow is intentionally Python/OpenCV based. Orientation classification is optional: when PaddleOCR is installed it can resolve 0/90/180/270 degree document orientation; otherwise the notebook uses image geometry and records the remaining ambiguity.

```python
from __future__ import annotations

from dataclasses import dataclass, field
from pathlib import Path
import json
import random
import time
from typing import Any

import cv2
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image, ImageOps
```

```python
SEED = 42
IMAGE_DIR = Path('../data/images')
RUN_DIR = Path('../runs/preprocessing')
EXPLICIT_IMAGE: Path | None = None

image_paths = sorted(p for p in IMAGE_DIR.glob('*') if p.suffix.lower() in {'.jpg', '.jpeg', '.png', '.tif', '.tiff', '.webp'})
if not image_paths:
    raise FileNotFoundError(f'No supported images found in {IMAGE_DIR.resolve()}')
rng = random.Random(SEED)
input_path = EXPLICIT_IMAGE or rng.choice(image_paths)
print(f'Selected: {input_path} ({len(image_paths)} images available)')
```

## Pipeline implementation

Page detection uses edges, morphology, and quadrilateral contours. A failed or low-confidence detection falls back to the complete image so preprocessing never silently clips text.

```python
@dataclass
class PreprocessConfig:
    proxy_max_side: int = 1400
    min_page_area_ratio: float = 0.25
    min_quad_score: float = 0.45
    deskew_limit_degrees: float = 10.0
    border_px: int = 12
    orientation: str = 'auto'  # auto, none, or 0/90/180/270
    output_size: tuple[int, int] | None = None

@dataclass
class PreprocessResult:
    source: str
    stages: dict[str, np.ndarray]
    metadata: dict[str, Any] = field(default_factory=dict)
    warnings: list[str] = field(default_factory=list)

def _resize_proxy(image: np.ndarray, max_side: int) -> tuple[np.ndarray, float]:
    h, w = image.shape[:2]
    scale = min(1.0, max_side / max(h, w))
    if scale == 1.0:
        return image.copy(), 1.0
    return cv2.resize(image, (round(w * scale), round(h * scale)), interpolation=cv2.INTER_AREA), scale

def load_image(path: Path) -> tuple[np.ndarray, dict[str, Any]]:
    with Image.open(path) as pil_image:
        exif_orientation = pil_image.getexif().get(274)
        pil_image = ImageOps.exif_transpose(pil_image).convert('RGB')
        rgb = np.asarray(pil_image)
    return cv2.cvtColor(rgb, cv2.COLOR_RGB2BGR), {'width': pil_image.width, 'height': pil_image.height, 'mode': pil_image.mode, 'exif_orientation': exif_orientation}

def _order_points(points: np.ndarray) -> np.ndarray:
    points = points.reshape(4, 2).astype('float32')
    sums, diffs = points.sum(axis=1), np.diff(points, axis=1).ravel()
    return np.array([points[np.argmin(sums)], points[np.argmin(diffs)], points[np.argmax(sums)], points[np.argmax(diffs)]], dtype='float32')

def _quad_score(quad: np.ndarray, shape: tuple[int, int], contour_area: float) -> float:
    h, w = shape
    area_ratio = contour_area / (h * w)
    sides = np.linalg.norm(np.roll(quad, -1, axis=0) - quad, axis=1)
    rectangularity = min(sides) / max(sides)
    return float(0.65 * min(1.0, area_ratio / 0.8) + 0.35 * rectangularity)

def detect_page(image: np.ndarray, config: PreprocessConfig) -> tuple[np.ndarray, dict[str, Any], list[str]]:
    proxy, scale = _resize_proxy(image, config.proxy_max_side)
    gray = cv2.cvtColor(proxy, cv2.COLOR_BGR2GRAY)
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)
    edges = cv2.Canny(blurred, 40, 140)
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (9, 9))
    closed = cv2.morphologyEx(edges, cv2.MORPH_CLOSE, kernel, iterations=2)
    contours, _ = cv2.findContours(closed, cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)
    best: tuple[float, np.ndarray] | None = None
    for contour in sorted(contours, key=cv2.contourArea, reverse=True)[:30]:
        area = cv2.contourArea(contour)
        if area / (proxy.shape[0] * proxy.shape[1]) < config.min_page_area_ratio:
            continue
        perimeter = cv2.arcLength(contour, True)
        approx = cv2.approxPolyDP(contour, 0.03 * perimeter, True)
        if len(approx) != 4 or not cv2.isContourConvex(approx):
            continue
        quad = _order_points(approx)
        score = _quad_score(quad, proxy.shape[:2], area)
        if best is None or score > best[0]:
            best = (score, quad)
    warnings: list[str] = []
    h, w = image.shape[:2]
    if best is None or best[0] < config.min_quad_score:
        warnings.append('No confident page quadrilateral found; using the full image.')
        return image.copy(), {'page_detected': False, 'page_score': None, 'corners': None, 'edge_proxy': closed}, warnings
    score, proxy_quad = best
    quad = proxy_quad / scale
    tl, tr, br, bl = quad
    width = int(max(np.linalg.norm(br - bl), np.linalg.norm(tr - tl)))
    height = int(max(np.linalg.norm(tr - br), np.linalg.norm(tl - bl)))
    if config.output_size:
        width, height = config.output_size
    destination = np.array([[0, 0], [width - 1, 0], [width - 1, height - 1], [0, height - 1]], dtype='float32')
    matrix = cv2.getPerspectiveTransform(quad, destination)
    warped = cv2.warpPerspective(image, matrix, (width, height), borderValue=(255, 255, 255))
    return warped, {'page_detected': True, 'page_score': score, 'corners': quad.tolist(), 'edge_proxy': closed}, warnings
```

```python
def estimate_orientation(image: np.ndarray, requested: str = 'auto') -> tuple[int, float, str]:
    if requested != 'auto':
        return int(requested) % 360, 1.0, 'manual'
    # Optional semantic classifier: it resolves the 180-degree ambiguity that
    # pure geometry cannot resolve. It is never required to run this notebook.
    try:
        from paddleocr import DocImgOrientationClassification
        result = next(iter(DocImgOrientationClassification(model_name='PP-LCNet_x1_0_doc_ori', device='cpu').predict(image, batch_size=1)))
        payload = result.json if isinstance(result.json, dict) else result.json()
        res = payload.get('res', payload)
        angle = int(res.get('label_names', [str(int(res.get('angle', 0)))])[0])
        score = float(np.asarray(res.get('scores', [1.0])).ravel()[0])
        return angle % 360, score, 'paddleocr'
    except Exception as exc:
        # Geometry can select portrait vs landscape, but not 0 vs 180.
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
        scores = []
        for angle in (0, 90, 180, 270):
            rotated = rotate_right(gray, angle)
            binary = cv2.threshold(rotated, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)[1]
            horizontal = cv2.morphologyEx(binary, cv2.MORPH_OPEN, cv2.getStructuringElement(cv2.MORPH_RECT, (max(15, rotated.shape[1] // 25), 1)))
            scores.append(float(horizontal.mean()))
        order = np.argsort(scores)[::-1]
        confidence = float((scores[order[0]] + 1e-6) / (scores[order[1]] + 1e-6)) if len(order) > 1 else 1.0
        return int((0, 90, 180, 270)[order[0]]), confidence, f'projection-fallback ({type(exc).__name__})'

def rotate_right(image: np.ndarray, angle: int) -> np.ndarray:
    return {0: image, 90: cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE), 180: cv2.rotate(image, cv2.ROTATE_180), 270: cv2.rotate(image, cv2.ROTATE_90_COUNTERCLOCKWISE)}[angle]

def deskew(image: np.ndarray, limit: float) -> tuple[np.ndarray, float]:
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    binary = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)[1]
    horizontal = cv2.morphologyEx(binary, cv2.MORPH_OPEN, cv2.getStructuringElement(cv2.MORPH_RECT, (max(20, image.shape[1] // 30), 1)))
    points = cv2.findNonZero(horizontal)
    if points is None or len(points) < 30:
        return image, 0.0
    angle = cv2.minAreaRect(points)[-1]
    if angle < -45:
        angle += 90
    if abs(angle) > limit:
        return image, 0.0
    h, w = image.shape[:2]
    matrix = cv2.getRotationMatrix2D((w / 2, h / 2), angle, 1.0)
    return cv2.warpAffine(image, matrix, (w, h), flags=cv2.INTER_CUBIC, borderMode=cv2.BORDER_CONSTANT, borderValue=(255, 255, 255)), float(angle)

def enhance_for_ocr(image: np.ndarray, border_px: int) -> tuple[np.ndarray, np.ndarray, np.ndarray]:
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    background = cv2.GaussianBlur(gray, (0, 0), sigmaX=25, sigmaY=25)
    normalized = cv2.divide(gray, background, scale=255)
    normalized = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8)).apply(normalized)
    normalized = cv2.medianBlur(normalized, 3)
    binary = cv2.adaptiveThreshold(normalized, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 31, 15)
    normalized = cv2.copyMakeBorder(normalized, border_px, border_px, border_px, border_px, cv2.BORDER_CONSTANT, value=255)
    binary = cv2.copyMakeBorder(binary, border_px, border_px, border_px, border_px, cv2.BORDER_CONSTANT, value=255)
    return normalized, binary, background

def preprocess_document(path: Path, config: PreprocessConfig = PreprocessConfig()) -> PreprocessResult:
    started = time.perf_counter()
    original, metadata = load_image(path)
    page, page_meta, warnings = detect_page(original, config)
    angle, orientation_score, orientation_method = estimate_orientation(page, config.orientation)
    oriented = rotate_right(page, angle)
    straightened, skew_angle = deskew(oriented, config.deskew_limit_degrees)
    normalized, binary, background = enhance_for_ocr(straightened, config.border_px)
    metadata.update({'orientation_angle': angle, 'orientation_score': orientation_score, 'orientation_method': orientation_method, 'deskew_angle': skew_angle, 'elapsed_seconds': time.perf_counter() - started, 'page_detected': page_meta['page_detected'], 'page_score': page_meta['page_score'], 'corners': page_meta['corners']})
    stages = {'original': original, 'page': page, 'oriented': oriented, 'deskewed': straightened, 'normalized_gray': normalized, 'adaptive_binary': binary, 'background_estimate': background, 'edge_proxy': page_meta['edge_proxy']}
    return PreprocessResult(str(path), stages, metadata, warnings)
```

```python
result = preprocess_document(input_path)
print(json.dumps({k: v for k, v in result.metadata.items() if k != 'corners'}, indent=2))
print('warnings:', result.warnings)

def show_stage_gallery(result: PreprocessResult) -> None:
    names = ['original', 'page', 'oriented', 'deskewed', 'normalized_gray', 'adaptive_binary']
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    for axis, name in zip(axes.flat, names):
        image = result.stages[name]
        if image.ndim == 3:
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        axis.imshow(image, cmap='gray')
        axis.set_title(name)
        axis.axis('off')
    plt.tight_layout()

show_stage_gallery(result)
```

```python
def save_result(result: PreprocessResult, root: Path = RUN_DIR) -> Path:
    output_dir = root / Path(result.source).stem
    output_dir.mkdir(parents=True, exist_ok=True)
    cv2.imwrite(str(output_dir / 'normalized_gray.png'), result.stages['normalized_gray'])
    cv2.imwrite(str(output_dir / 'adaptive_binary.png'), result.stages['adaptive_binary'])
    (output_dir / 'metadata.json').write_text(json.dumps({'source': result.source, 'metadata': result.metadata, 'warnings': result.warnings}, indent=2), encoding='utf-8')
    return output_dir

save_result(result)
```

## Batch check

The batch run is diagnostic rather than a claim of OCR accuracy. Review the generated galleries and metadata, especially when page detection is low-confidence or the optional orientation model is unavailable.

```python
batch = []
for path in image_paths:
    item = preprocess_document(path)
    output_dir = save_result(item)
    batch.append({'image': path.name, 'output': str(output_dir), **{k: item.metadata[k] for k in ('orientation_angle', 'orientation_score', 'orientation_method', 'deskew_angle', 'page_detected', 'page_score')}, 'warnings': item.warnings})

print(json.dumps(batch, indent=2))
```

### Interpretation and next steps

Use `normalized_gray.png` as the default OCR input and compare it with `adaptive_binary.png`. Add a small labelled validation set with expected orientation, page corners, and anchor text before tuning thresholds. If the camera images routinely contain curved pages or severe perspective, enable PaddleOCR's document preprocessor/unwarping model rather than increasing classical morphology.
